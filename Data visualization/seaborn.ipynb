```python
# Seaborn
# Seaborn is a Python data visualization library based on matplotlib.
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.get_dataset_names()
```

```python
sns.set_theme(style="ticks")

tips=sns.load_dataset("tips")

sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker",
    size="size",
    style="smoker"
)
```

```python

x_vals=[2,4,6,8,10,12,14,16]
x_sq=[i*i for i in x_vals]
sns.set_theme(style="ticks")
sns.relplot(
    x=x_vals,
    y=x_sq,
    kind="line"
)
```

```python
#scatter plot
tips=sns.load_dataset("tips")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time"
)
```

```python
#line plot
flights=sns.load_dataset("flights")
sns.lineplot(
    data=flights,
    x="year",
    y="passengers"
)
```

```python
#categorical plots

# print(tips.head())
sns.barplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)
```

```python
#box plot

sns.boxplot(
    data=tips,
    x="day",
    y="tip",

)
```

```python
#distribution plots
peng=sns.load_dataset("penguins")

sns.histplot(
    data=peng,
    x="body_mass_g",
    bins=20
)
```

```python
#matrix plots--- eg. heatmaps
# flights.head()

#pivot the data
flights_pivot=flights.pivot(index="month", columns="year", values="passengers")
print(flights_pivot)


sns.heatmap(
    flights_pivot,
    cmap="coolwarm",
    annot=True,
    fmt="d"
)
plt.title("passengers heatmap")
plt.tight_layout()
```

```python
#best practices
# use matplotlib and seaborn together. create fig and ax using matplot
# create individual plots using seaborn
fig,ax=plt.subplots()

sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    ax=ax,
    marker="o",
    errorbar=None # hide confidence interval
)
ax.set_title("avg bill for each day")
ax.set_xlabel("Days")
ax.set_ylabel("Total bill amount")

fig.tight_layout()
```
