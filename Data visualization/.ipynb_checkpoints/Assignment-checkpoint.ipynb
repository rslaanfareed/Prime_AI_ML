```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


#Question 1 - Histogram
np.random.seed(0)
scores=np.random.normal(loc=70, scale=10, size=200)
plt.hist(scores, color="skyblue",edgecolor="black",bins=20)

plt.xlabel("Score")
plt.ylabel("Frequency")
plt.title("Distribution of Scores")

plt.axvline(
    x=70,
    label="Mean (70)",
    color="red",
    linewidth="2",
    linestyle="--"
)

plt.show()
```

```python
#Question 3
tips=sns.load_dataset("tips")
sns.boxplot(
    data=tips,
    x="total_bill",
    y="day",
    hue="sex"
)
plt.title("Total Bill Distribution by Day and Gender")
plt.xlabel("Day of the Week")
plt.ylabel("Total Bill ($)")
```

```python
#Question 4
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    style="smoker",
    s=100
)
plt.title("Total bill vs Tips")
plt.xlabel("Total Bill")
plt.ylabel("Tip")

plt.legend()
```

```python
#Question 5
pivot_tips=tips.pivot_table(index="day", columns="time", values="tip",aggfunc="mean",observed=False)

sns.heatmap(pivot_tips, annot=True, cmap="coolwarm", fmt=".2f", cbar_kws={'label': 'Average Tip ($)'})

# 4. Add title and axis labels
plt.title("Average Tip ($) by Day and Time", fontsize=14, pad=15)
plt.xlabel("Time of Day", fontsize=12)
plt.ylabel("Day of the Week", fontsize=12)
```
