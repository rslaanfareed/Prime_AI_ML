```python
# real data
oscar_winners=[
    "The Dark knight",
    "The Heart locker",
    "The king's Speech",
    "The Artist",
    "Argo"
]
years=[2008,2009,2010,2011,2012]
oscar_revenue=[1005,170,427,133,232] #in $M

#plot
plt.plot(years,oscar_revenue)
plt.title("Oscar Movies Revenue in each year")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")

plt.show()
```

```python
# non_oscar winner movies
non_oscar_movies=["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue=[378,2788,829,185,275] #in $M

#format string
#fmt='[marker][line][color]'
# "o-b"      Blue circles + solid line
# "s--r"     Red squares + dashed line
# "^-g"      Green triangles + solid line
# "x:k"      Black x markers + dotted line
# "*-.m"     Magenta stars + dash-dot line


#plot
plt.plot(years,oscar_revenue,"o-b",label="Oscar Movies")

#we cal also pass values like this: 
#plt.plot(years, oscar_revenue, color=red, marker="o", linestyle="--", linewidth=3, label="Oscar Movies")
# we can also pass hex values if we want

plt.plot(years,non_oscar_revenue,"o-r",label="Non Oscar Movies")
plt.title("Oscar Movies vs Non Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend( ) #to show the label box of both the lines

plt.show()
```

```python
import numpy as np

# Bar chart: compare two sets of data
oscar_movies=[
    "The Dark knight",
    "The Heart locker",
    "The king's Speech",
    "The Artist",
    "Argo"
]
oscar_revenue=[1005,170,427,133,232] #in $M

non_oscar_movies=["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue=[378,2788,829,185,275] #in $M

years=[2008,2009,2010,2011,2012]

x=np.arange(len(years))
width=0.4

# x-width/2=x-0.2
# x+width/2=x+0.2

plt.bar(x-width/2,oscar_revenue,width,label="Oscar Movies") #shifting to left
plt.bar(x+width/2,non_oscar_revenue,width,label="Non Oscar Movies") #shifting to right
plt.title("Oscar vs Non oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue in $M")

plt.legend()

#replace x values(1,2,3..) with year values now
plt.xticks(x,years)

plt.show()
```

```python
#horizontal bar charts

oscar_movies=[
    "The Dark knight",
    "The Heart locker",
    "The king's Speech",
    "The Artist",
    "Argo"
]
oscar_revenue=[1005,170,427,133,232] #in $M

years=[2008,2009,2010,2011,2012]

plt.barh(oscar_movies,oscar_revenue)
plt.xlabel("revenue")
plt.ylabel("movie names")

plt.show()
```

```python
# Scatter Plots
# A scatter plot shows the relationship between two variables by plotting points on a 2D plane.

people = ["Person A", "Person B", "Person C", "Person D", "Person E",
"Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

#customizations
# color, size, marker, transparency, colormap

colors=["green" if x<=130 else "red" for x in bp]
#plt.scatter(..,..,.., color=colors)



#colormap
#plt.scatter(..,..,.., cmap="plasma",c=bp)

plt.scatter(age,bp,   cmap="OrRd",c=bp,s=bp,marker="o",alpha=.8) #alpha = transparency. default 1
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
plt.colorbar(label="BP")

# annotations
# plt.annotate( "text", xy=(x_point, y_point), xytext=(x_text, y_text) )
for i in range(len(people)):
    plt.annotate(people[i],xy=(age[i],bp[i]), xytext=(age[i]+2,bp[i]-0.5))

plt.xlim(min(age),max(age)+10)
plt.show()
```
