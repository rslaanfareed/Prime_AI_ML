# Clustering

## What is Clustering?
Clustering is an **unsupervised ML technique** that automatically groups **similar, unlabeled data points** into clusters based on defined similarity metrics.



## Main Idea
- No target labels are given.
- Similar data points are placed in the same cluster.
- Different groups form based on similarity.

## Types of Clustering

### 1. Partitioning-Based
Divides data into a fixed number of clusters.

**Example:** K-Means Clustering

### 2. Hierarchical Clustering
Builds a hierarchy of clusters by progressively grouping similar data points.

### 3. Density-Based
Forms clusters based on regions of high data-point density and can identify isolated points.

**Example:** DBSCAN

## Quick Summary

| Type | Example | Main Idea |
|---|---|---|
| Partitioning-based | K-Means | Divide data into clusters |
| Hierarchical | Hierarchical Clustering | Build a cluster hierarchy |
| Density-based | DBSCAN | Find dense regions and isolated points |

**Key point:** Clustering finds natural groups in **unlabeled data**.
