# Unsupervised Machine Learning

## What is it?
Unsupervised ML learns from **unlabeled data**.  
There is no target/output column; the algorithm tries to discover the hidden structure in the data.

## Why is it used?
To find useful information or structure when labeled data is unavailable or expensive to obtain.

## Main Purposes

- **Pattern discovery:** Find meaningful patterns or relationships in data.
- **Clustering:** Group similar data points together.
  - Example: Group customers based on purchasing behavior.
- **Anomaly detection:** Identify unusual or abnormal observations.
  - Example: Detect unusual credit-card transactions.
- **Dimensionality reduction:** Reduce the number of features while preserving important information.
  - Example: PCA for visualization or preprocessing.
- **Association discovery:** Find items or features that frequently occur together.
  - Example: Market basket analysis: customers who buy X often buy Y.

## Common Algorithms

- **K-Means** → Clustering
- **Hierarchical Clustering** → Clustering
- **DBSCAN** → Clustering + anomaly detection
- **PCA** → Dimensionality reduction
- **Apriori** → Association rule mining

## Examples

| Task | Example |
|---|---|
| Customer segmentation | Group customers by behavior |
| Anomaly detection | Find fraudulent transactions |
| Pattern recognition | Discover groups/patterns in user behavior |
| Recommendation | Find products commonly associated with each other |
| Visualization | Reduce high-dimensional data using PCA |

## Benefits

- Works without labeled data.
- Helps discover hidden patterns.
- Useful for exploratory data analysis.
- Can reveal anomalies and unusual behavior.
- Can simplify high-dimensional datasets.

## Key Idea

**Supervised ML:** learns from labeled data → predicts an output.  
**Unsupervised ML:** learns from unlabeled data → discovers structure, patterns, groups, or anomalies.
