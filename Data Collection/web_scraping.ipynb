```python
with open("scraped_data/data1.html","w",encoding="utf-8") as f:
    f.write(res.text)
```

```python
from bs4 import BeautifulSoup
with open("scraped_data/data1.html","r") as f:
    html_content=f.read()

soup=BeautifulSoup(html_content,"lxml")
```

```python
all_h3=soup.find_all("h3")
all_countries=[]
for h3 in all_h3:
    # get text
    country_name=h3.get_text(strip=True)
    # print(country_name)
    # find parent
    country_class=h3.find_parent("div")["class"]
    # print(country_class)

    #find next, find previous
    population=h3.find_next("div").select("span.country-population")[0].get_text(strip=True)

    # print(h3.attrs)

    #APPEND IN THE LIST
    all_countries.append([country_name,population])
```

```python
#CREATE DATAFRAME
import pandas as pd
df=pd.DataFrame(all_countries,columns=["Name","Population"])
```

```python
#STORE IN CSV
df.to_csv("cleaned_data/data.csv",index=False)
```
