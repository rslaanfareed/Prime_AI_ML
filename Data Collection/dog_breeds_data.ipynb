```python
import requests
url="https://dogapi.dog/api/v2/breeds"
res=requests.get(url)
```

```python
if res.status_code==200:
    print(f"Success. Status Code: {res.status_code}")
```

```python
# res.headers["Content-type"]

#converting to json
json_data=res.json()
```

```python
#chek types
print(type(json_data))
print(json_data)
```

```python
#converting to dataframe
import pandas as pd
df=pd.json_normalize(json_data["data"])
```

```python
df2=df.copy()
```

```python
## Inspect the Data
# df.head()
# df.dtypes
# df.shape
# df.columns
# df.isnull().sum()
# df.duplicated().sum()
```

```python
#rename columns
df.rename(columns={
    "attributes.name": "name",
    "attributes.life.min": "life_min",
    "attributes.life.max": "life_max",
    "attributes.hypoallergenic": "hypoallergenic"
    "attributes.male_weight.min": "male_weight_min",
    "attributes.male_weight.max": "male_weight_max",
    "attributes.female_weight.min": "female_weight_min",
    "attributes.female_weight.max": "female_weight_max",
    "attributes.hypoallergenic": "hypoallergenic",
    "relationships.group.data.id": "group_id",
    "relationships.group.data.type": "group_type"
},inplace=True)
```

```python
#create avg life col
df["Avg_life"]=(df["life_min"]+df["life_max"])/2
```

```python
df.head()
df.nunique()
```

```python
# Drop columns with no analytical value
df.drop(columns=[
    "id",
    "type",
    "group_id",
    "group_type",
    "attributes.description"
], inplace=True)
```

```python
df.head()
```

```python
df["name"] = df["name"].str.strip()
df["name"] = df["name"].str.lower()
df.sort_values("name")
```

```python
df.to_csv("cleaned_data/dog_breeds_cleaned.csv", index=False)
```
