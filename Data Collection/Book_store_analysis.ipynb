```python
import requests
url="https://books.toscrape.com/"
res=requests.get(url)
```

```python
res.status_code
```

```python
with open("scraped_data/book.html","w",encoding="utf-8") as f:
    f.write(res.text)
```

```python
from bs4 import BeautifulSoup
with open("scraped_data/book.html","r") as f:
    html_content=f.read()
soup=BeautifulSoup(html_content,"html.parser")
```

```python
# Find all books
books = soup.find_all("article", class_="product_pod")

all_books = []

# Rating conversion dictionary
rating_map = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

for book in books:

    # Title
    title = book.find("h3").find("a").get("title")

    # Price
    price = book.find("p", class_="price_color").get_text(strip=True)

    # Remove the pound sign and encoding characters
    price = price.replace("£", "")
    price = price.replace("Ã‚Â", "")
    price = float(price)

    # Availability
    availability = book.find("p",class_="instock availability").get_text(strip=True)

    # Rating
    rating = book.find("p", class_="star-rating").get("class")[1]
    rating = rating_map[rating]

    # Store data
    all_books.append({
        "Title": title,
        "Price": price,
        "Availability": availability,
        "Rating": rating
    })
    
```

```python
# Create DataFrame
import pandas as pd
df = pd.DataFrame(all_books)

df.head()
```

```python
# sort 
df.sort_values("Title",inplace=True)
df.reset_index(inplace=True)
```

```python
df.drop(columns="index",inplace=True)
```
