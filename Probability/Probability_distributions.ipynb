```python
# Binomial Distribution - A fair coin is tossed 5 times. what is the probability that exactly 3 heads are obtained?
from scipy.stats import binom

n = 5 # no of trials
k = 3 # no of successes
p = 0.5 # prob of getting head

prob = binom.pmf(k, n, p)
print(prob)   # 5/16
```

```python
# Uniform Distribution
import numpy as np
import matplotlib.pyplot as plt

# 10_00_000 random values between 0 and 10
values = np.random.uniform(0, 10, 10_00_000)

# plotting a histogram - becomes more uniform with more values
plt.hist(values, bins=100, density=True, alpha =0.3)

plt.title("Continuous Uniform Distribution between 0 and 10")
plt.xlabel("x")
plt.ylabel("Probability Density")
plt.grid(True)

plt.show()
```

```python
#Normal Distribution

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parameters
mu = 70       #mean
sigma = 10    #std dev

# X-axis values
x = np. linspace(30, 110, 1000)

# Normal PDF
y = norm.pdf(x, mu, sigma)

# Plot the bell curve
plt.plot(x, y, color='black', linewidth=2, label='Normal Distribution')

# Shade 68% region (mu ± 1 sigma)
x1 = np. linspace(mu - sigma, mu + sigma, 1000)
plt.fill_between(x1, norm.pdf(x1, mu, sigma), color='green', alpha=0.3, label='68% region (1 sigma)')

# Shade 95% region (p ± 2o)
x2 = np. linspace(mu - 2*sigma, mu + 2*sigma, 1000)
plt.fill_between(x2, norm.pdf(x2, mu, sigma), color='yellow', alpha=0.2, label='95% region (2 sigma)')

# Shade 99.7% region (p ± 3o)
x3 = np. linspace(mu - 3*sigma, mu + 3*sigma, 1000)
plt.fill_between(x3, norm.pdf(x3, mu, sigma), color='red', alpha=0.1, label='99.7% region (3 sigma)')

# Labels and title
plt.title("Normal Distribution of Student Scores")
plt.xlabel("Score")
plt.ylabel("Probability Density")
plt. legend()
plt.show()
```
