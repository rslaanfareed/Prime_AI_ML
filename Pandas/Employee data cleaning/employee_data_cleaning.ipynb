```python
# create a copy of the data
employees_copy=employees.copy()
```

```python
employees.shape
employees.dtypes
employees.nunique()
employees.describe()
```

```python
#fill missing department ffill
employees["Department"]=employees["Department"].ffill()
#fill missing age with mean
employees["Age"]=employees["Age"].fillna(employees["Age"].mean())
#Fill missing Salary with average salary.
employees["Salary"]=employees["Salary"].fillna(employees["Salary"].mean())
#Fill missing City using ffill.
employees["City"]=employees["City"].ffill()
```

```python
#dropping duplicates
employees = employees.drop_duplicates(subset=['EmployeeID'])
```

```python
#Verify duplicates are removed.
employees.duplicated()
```

```python
#remove leading & trailing spaces from name,city
employees["Name"]=employees["Name"].str.strip()
employees["City"]=employees["City"].str.strip()
```

```python
# Create Bonus
employees["Bonus"]=employees['Salary']*0.1
```

```python
employees["Total Salary"]=employees["Salary"]+employees["Bonus"]
```

```python
#Create age groups
# 1. Define the age grouping logic
def categorize_age(age):
    if age < 18:
        return 'Child'
    elif age < 35:
        return 'Young Adult'
    elif age < 55:
        return 'Middle Aged'
    else:
        return 'Senior'

# 2. Apply the function to the 'Age' column
employees['age_group'] = employees['Age'].apply(categorize_age)
```

```python
#map performance to numbers
performance_dict = {
    'Poor':1,
    'Average':2,
    'Good':3,
    'Excellent':4
}

employees['Performance'] = employees['Performance'].map(performance_dict)
```

```python
employees["Department"]=employees["Department"].replace("HR","Human Resources")
employees["Department"]=employees["Department"].replace("IT","Information Tech")
```

```python
#Find average salary department-wise
employees.groupby("Department")["Salary"].mean()
```

```python
# Highest salary department-wise
employees.groupby("Department")["Salary"].max()
```

```python
# Employee count department-wise
employees.groupby("Department")["EmployeeID"].count()
```
