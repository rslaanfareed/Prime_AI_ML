```python
#Question 1
#load IRIS datased 
import pandas as pd
df=pd.read_csv("IRIS.csv")

#A

# first 10 rows
df.head(10)
#shape
df.shape
#data types
df.dtypes
df.describe()

#B

#petal length > 4.5
df[df["petal_length"]>4.5]
#species = "Iris-virginica"
df[df["species"]=="Iris-virginica"]

#C

#compute avg sepal length
df.groupby("species").agg(avg_sepal_length=("sepal_length","mean"))
#max petal witdth
df.groupby("species").agg(max_petal_width=("petal_width","max"))
#std dev of sepal width
df.groupby("species").agg(std_dev_sepal_w=("sepal_width","std"))

#D

#new column petal ratio
df=df.assign(petal_ratio=df["petal_length"]/df["petal_width"])
#avg petal ratio for each specie
df.groupby("species").agg(avg_petal_ratio=("petal_ratio","mean"))
```

```python
#load TITANIC dataset
df=pd.read_csv("Titanic-Dataset.csv")

#A -  Display only columns: A. Name, Sex, Age, Fare, Survived
df[["Name","Sex","Age","Fare","Survived"]]

#B - Select passengers who:
# Are female and
# Fare > 30

df[(df["Sex"]=="female") & (df["Fare"]>30)]

#C - group by Pclass and ocmpute
# survival rate, average fare, average age

df.groupby("Pclass").agg(
    avg_fare=("Fare","mean"),
    avg_age=("Age","mean"),
    survival_rate=("Survived","mean")
)


```
