# Random Forest

Random Forest is a **Bagging-based ensemble learning method** that combines multiple **Decision Trees (DTs)** to produce a stronger and more stable model.

The main idea is to create different datasets from the original dataset, train a different Decision Tree on each dataset, and then combine their predictions.

---

## Why Random Forest?

A single Decision Tree can easily **overfit** the training data and have high variance.

Random Forest reduces this problem by using **multiple Decision Trees** instead of relying on one tree.

```text
One Decision Tree
       ↓
   Can overfit
       ↓
Random Forest
       ↓
Multiple Decision Trees
       ↓
Combined Prediction
       ↓
More stable model
```

---

# Complete Random Forest Process



## Step 1: Start with the Dataset

We start with a dataset containing:

- `n` samples (rows)
- `f` features (columns)

```text
Dataset
    ↓
n samples
f features
```

---

## Step 2: Row Sampling

Random Forest uses **bootstrap sampling**.  

We randomly select rows from the original dataset **with replacement**.

### Bootstrap Sampling with Replacement:

Bootstrap sampling means creating a new dataset by randomly selecting samples from the original dataset with replacement.  

With replacement means that after selecting an observation, we put it back into the dataset before selecting the next one.  

This creates different datasets for different trees.  
`
Example:  
`

Original Dataset:  
[A, B, C, D, E]  

Bootstrap Sample:  
[B, D, B, E, A]  

```text
Original Dataset
       ↓
Bootstrap Sampling
       ↓
 ┌────────┬────────┬────────┐
 ↓        ↓        ↓
D1       D2       D3
 ↓        ↓        ↓
Tree 1   Tree 2   Tree 3
```

Because sampling is done **with replacement**, a row can appear more than once in a sampled dataset.

Some rows may not appear in a particular sample at all.

---

## Step 3: Feature Sampling

Random Forest also randomly selects a subset of features for each tree.

For example, suppose the original dataset has:

```text
f1, f2, f3, f4, f5, f6
```

If the maximum number of features is `2`, one tree might use:

```text
Tree 1 → f1, f3
```

Another tree might use:

```text
Tree 2 → f2, f5
```

Another tree might use:

```text
Tree 3 → f4, f6
```

This is called **feature sampling** or **feature bagging**.

The purpose is to make the trees more different from each other.

---

## Step 4: Create Different Training Datasets

Now each tree gets a different combination of:

- Randomly sampled rows
- Randomly selected features

For example:

```text
D1 → n sampled rows + selected features → Tree 1

D2 → n sampled rows + selected features → Tree 2

D3 → n sampled rows + selected features → Tree 3
```

Therefore, the trees do not all learn exactly the same thing.

---

## Step 5: Train the Decision Trees

A Decision Tree is trained on each sampled dataset.

```text
D1 → Decision Tree 1
D2 → Decision Tree 2
D3 → Decision Tree 3
```

The trees are trained **independently**, so they can be trained in parallel.

```text
           ┌→ Tree 1
Dataset ───┼→ Tree 2
           └→ Tree 3
```

Each tree learns a slightly different pattern because it receives different rows and features.

---

# Step 6: Make Predictions

When a new data point arrives, every Decision Tree makes its own prediction.

For example:

```text
New Data
   ↓
 ┌───────────────┐
 ↓       ↓       ↓
Tree 1  Tree 2  Tree 3
 ↓       ↓       ↓
 Yes     No      Yes
```

Now we have multiple individual predictions.

---

# Step 7: Aggregate the Predictions

The predictions from all the trees are combined.

### Classification

For classification, Random Forest usually uses **majority voting**.

```text
Tree 1 → Yes
Tree 2 → No
Tree 3 → Yes

Yes → 2 votes
No  → 1 vote

Final Prediction → Yes
```

### Regression

For regression, the predictions are generally combined using their **average**.

```text
Tree 1 → 20
Tree 2 → 24
Tree 3 → 22

Final Prediction → Average
```

---

# Complete Random Forest Workflow

```text
Original Dataset
       ↓
Row Sampling
(Bootstrap Sampling)
       ↓
Feature Sampling
(Feature Bagging)
       ↓
 ┌────────┬────────┬────────┐
 ↓        ↓        ↓
D1       D2       D3
 ↓        ↓        ↓
Tree 1   Tree 2   Tree 3
 ↓        ↓        ↓
Prediction Prediction Prediction
 └────────┬────────┘
          ↓
     Aggregation
          ↓
 Classification → Majority Voting
 Regression     → Average
          ↓
    Final Prediction
```

---

# Why Use Both Row and Feature Sampling?

Random Forest uses randomness in **two ways**:

### 1. Random Rows

Bootstrap sampling creates different training datasets.

```text
Different rows
     ↓
Different training samples
     ↓
Different trees
```

### 2. Random Features

Feature sampling prevents every tree from relying on the same features.

```text
Different features
     ↓
Different tree structures
     ↓
More diverse trees
```

Together, these make the trees **less correlated** with each other.

When their predictions are combined, the overall model becomes more stable.

---

# Random Forest in One Sentence

> **Random Forest creates many Decision Trees using random rows and random features, then combines their predictions to produce the final result.**

---

## Key Points

- Random Forest is a **Bagging-based ensemble method**.
- Its base learners are **Decision Trees**.
- It uses **bootstrap sampling** for rows.
- Bootstrap sampling is done **with replacement**.
- It also performs **feature sampling**.
- Each tree receives a different combination of data and features.
- Trees are trained **independently**.
- **Classification → Majority Voting**
- **Regression → Average**
- The randomness makes the trees more diverse.
- Combining many trees gives a **more stable model** than relying on a single Decision Tree.


# Implementation

````python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

````

````python
titanic = sns.load_dataset("titanic")

features = ["pclass", "sex", "fare", "embarked", "age"]
target = ["survived"]

````

````python
# handle missing data
imp_median = SimpleImputer(strategy="median")
titanic[["age"]] = imp_median.fit_transform(titanic[["age"]])

imp_freq = SimpleImputer(strategy="most_frequent")
titanic[["embarked"]] = imp_freq.fit_transform(titanic[["embarked"]])

````

````python
# encode
le = LabelEncoder()

titanic["sex"] = le.fit_transform(titanic["sex"])
titanic["embarked"] = le.fit_transform(titanic["embarked"])

X = titanic[features]
y = titanic["survived"]

# Train Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

````

````python
# Decision Tree
model = DecisionTreeClassifier(max_depth=5)

model.fit(X_train, y_train)

y_pred_test = model.predict(X_test)
y_pred_train = model.predict(X_train)

print("Training accuracy: ", accuracy_score(y_train, y_pred_train) * 100, "%")
print("Testing accuracy: ", accuracy_score(y_test, y_pred_test) * 100, "%")

# classic case of overfitting
````

````python

from sklearn.tree import plot_tree

plt.figure(figsize=(18, 10))

plot_tree(
    model,
    feature_names=X.columns,
    class_names=["Died", "Survived"],
    filled=True
)

plt.tight_layout()
````

````python
# random forest
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=299,
    oob_score=True,
    max_depth=5
)

rf.fit(X_train, y_train)

y_pred = rf.predict(X_test)

print("OOB score: ", rf.oob_score_)
print("testing accuracy: ", accuracy_score(y_test, y_pred))
````
