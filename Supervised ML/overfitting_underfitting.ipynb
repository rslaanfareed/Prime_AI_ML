# What is Overfitting?

**Overfitting means the model memorizes the training data instead of learning the real pattern.**

Imagine these white dots are your training data.

The orange curve represents the **actual relationship**.

The blue wiggly curve tries to pass through **every single point**, including random noise.

Instead of learning the trend, it learns the mistakes.





---

## Why does it happen?

The model is **too complex**.

For example:

* Polynomial of degree 15 instead of degree 2
* Very deep decision tree
* Neural network with too many parameters

A complex model has enough flexibility to fit almost anything.

---

### "Fit → X → Y"

Remember what fitting means.

Training data consists of

```
X → Input features
Y → Correct outputs
```

During training, the algorithm tries to learn a function



so that predictions become as close as possible to the real outputs.

---

### Why is J(θ)=0?


Remember


J(θ)


is the **cost function**.

For training data,



If the model passes through every training point

```
Prediction = Actual
```

then

```
Error = 0
```

for every sample.

Therefore



and

J(θ)=0

This looks amazing...

But only on the training data.

---

### Why is training accuracy high?


```
Training = 95%
```

or maybe even 100%.

Since the model memorized every point,

it predicts the training examples almost perfectly.

So

```
Training accuracy ↑
Training error ↓
```

---

### Why is test accuracy low?

Suppose the model learned this weird curve.

Now a new point appears.



The model asks

"I've never seen this point."

Since it memorized instead of understanding,

its prediction is poor.

Therefore

```
Test accuracy ↓
```

Example:

```
Training Accuracy = 99%

Testing Accuracy = 63%
```

This large gap is a classic sign of overfitting.

---

### Low Bias

Bias means

> Error caused by making overly simple assumptions.

Since an overfitted model is extremely flexible,

it makes very few assumptions.

So


```
Low Bias
```


> errors from wrong assumptions

Because the model isn't assuming a simple relationship, this source of error is small.

---

### High Variance

Variance means

> How much the model changes when the training data changes slightly.



A tiny data change completely changes the learned curve.

That's **high variance**.

---

### "Sensitive to small data changes"

This sentence is describing variance.

High variance means

```
Small change in training data
↓

Very different model
↓

Different predictions
```

The model is unstable.

---

## The Bias-Variance relationship

| Underfitting        | Good Fit | Overfitting   |
| ------------------- | -------- | ------------- |
| High Bias           | Balanced | Low Bias      |
| Low Variance        | Balanced | High Variance |
| Training error High | Low      | Very Low      |
| Test error High     | Low      | High          |

---


### How to recognize overfitting

You usually see something like:

```
Training Accuracy : 98%

Testing Accuracy : 65%
```

or

```
Training Loss : Very Low

Validation Loss : High
```

Large difference between training and testing performance usually indicates overfitting.

# Underfitting

**Definition:**
The model is **too simple** to learn the actual pattern in the data.

### Characteristics

* **Training accuracy:** Low (e.g., 70%)
* **Testing accuracy:** Also low (e.g., 65%)
* **High Bias:** Makes overly simple assumptions, so it misses the true relationship.
* **Low Variance:** Small changes in the training data don't change the model much.

### Graph Intuition



The blue line is too simple to capture the curved relationship, so it performs poorly on both training and test data.

### Why does it happen?

* Model is too simple (e.g., linear regression for nonlinear data)
* Too few features
* Too much regularization
* Not trained enough

### How to identify it?

* Low training accuracy
* Low testing accuracy
* Both errors are high

### How to fix it?

* Use a more complex model
* Add better features
* Reduce regularization
* Train longer (if applicable)

### Quick comparison

| Underfitting       | Overfitting                |
| ------------------ | -------------------------- |
| Model too simple   | Model too complex          |
| High Bias          | Low Bias                   |
| Low Variance       | High Variance              |
| Low train accuracy | Very high train accuracy   |
| Low test accuracy  | Low test accuracy          |
| Misses the pattern | Learns the pattern + noise |


## How to Fix Underfitting & Overfitting

### Underfitting (High Bias)

**Problem:** Model is too simple.

**Fixes:**

* Increase model complexity.

  * More parameters
  * More layers (Neural Networks)
  * More neurons per layer
* Decrease regularization (L1/L2).
* Add more useful features.
* More training data **usually does not help** because the model is too simple.

---

### Overfitting (High Variance)

**Problem:** Model is too complex.

**Fixes:**

* Decrease model complexity.

  * Fewer parameters
  * Fewer layers
  * Smaller network
* Increase regularization (L1/L2).
* Add more training data (helps the model generalize).
* Use techniques like Early Stopping or Dropout (for neural networks).

---

## Bias-Variance Tradeoff

The goal is **not** to minimize only bias or only variance.

* High Bias → Underfitting
* High Variance → Overfitting

The goal is to find the **sweet spot** where:

* Bias is low enough to learn the pattern.
* Variance is low enough to generalize to new data.

This balance is called the **Bias-Variance Tradeoff**.

### Quick Summary

| Underfitting                   | Overfitting               |
| ------------------------------ | ------------------------- |
| High Bias                      | High Variance             |
| Increase model complexity      | Decrease model complexity |
| Decrease regularization        | Increase regularization   |
| Add more features              | Add more data             |
| More data usually doesn't help | More data often helps     |
