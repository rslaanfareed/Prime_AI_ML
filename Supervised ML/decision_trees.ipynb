# Decision Trees

A **Decision Tree** is a supervised machine learning algorithm that makes decisions by asking a series of **yes/no questions** about the data.

It looks like an upside-down tree:
- **Root Node** → Starting point
- **Decision Nodes** → Questions based on features
- **Leaf Nodes** → Final prediction



---

# Goal of a Decision Tree

The objective is simple:

> **Split the data into groups that are as "pure" as possible.**

Instead of memorizing data, the tree keeps asking questions that separate similar examples together.

Example:

```
Is Age < 30?
        │
   Yes      No
   │         │
Buy?      Income > 50k?
```



---

# How Does It Make Decisions?

At every node, the tree:

1. Looks at every feature.
2. Tries many possible splits.
3. Chooses the split that creates the **purest child nodes**.
4. Repeats this process until a stopping condition is met.

It is a **greedy algorithm** because it chooses the best split **at the current step**, not necessarily the best overall tree.

---

# Pure vs Impure Nodes




## Pure Node

A node is **pure** when **all samples belong to the same class**.

Example:

| Student | Pass |
|---------|------|
| A | Yes |
| B | Yes |
| C | Yes |

Pure (100% Yes)

---

## Impure Node

A node is **impure** when it contains a mixture of classes.

Example:

| Student | Pass |
|---------|------|
| A | Yes |
| B | No |
| C | Yes |

Impure (mixed Yes and No)


---

# How Does the Tree Measure Purity?

The two most common measures are:

- **Gini Impurity** (used by CART)
- **Entropy / Information Gain** (used by ID3, C4.5)

Lower impurity = Better split.

The algorithm always chooses the split that **reduces impurity the most**.

---

# When Does the Tree Stop Growing?

The tree stops when:

- Every node is pure.
- Maximum tree depth is reached.
- Too few samples remain to split.
- No split improves purity.

---

# Prediction Process

For a new data point:

1. Start at the root.
2. Answer each question.
3. Follow the corresponding branch.
4. Stop at a leaf node.
5. Output the class (classification) or value (regression).

---

# Key Intuition

Think of a decision tree like a game of **20 Questions**.

Each question is chosen to eliminate as much uncertainty as possible until only one answer remains.

# Entropy

Entropy measures the **impurity**, **disorder**, or **uncertainty** in a dataset.

A Decision Tree chooses the split that **reduces entropy the most**.

---

## Intuition

Think of entropy as **confusion**.

- Only one class → No confusion → Low entropy
- Classes equally mixed → Maximum confusion → High entropy

Example:

```text
[Yes, Yes, Yes, Yes]
```

Entropy = **0** (Pure)

```text
[Yes, No, Yes, No]
```

Entropy = **1** (Maximum impurity)

---

# Entropy Formula

For **c classes**:

$$
E(S) = -\sum_{i=1}^{c} p_i \log_2(p_i)
$$

For **binary classification**:

$$
E(S)= -P(\text{Yes})\log_2(P(\text{Yes}))
      -P(\text{No})\log_2(P(\text{No}))
$$

---

# Understanding the Formula

## $S$

The current dataset (or node).

---

## $c$

Number of classes.

Example:

```text
Yes / No
```

$$
c=2
$$

---

## $p_i$

Probability of class $i$.

Formula:

$$
p_i=\frac{\text{Number of samples in class }i}
{\text{Total number of samples}}
$$

or

$$
p_i=\frac{n_i}{N}
$$

Where:

- $n_i$ = Number of samples belonging to class $i$
- $N$ = Total number of samples

---

## Example: Calculating $p_i$

Dataset:

```text
[3 Yes, 6 No]
```

Total samples:

$$
N=3+6=9
$$

Probabilities:

$$
P(\text{Yes})=\frac{3}{9}=0.333
$$

$$
P(\text{No})=\frac{6}{9}=0.667
$$

The probabilities always satisfy:

$$
\sum_{i=1}^{c} p_i = 1
$$

---

## Why $\log_2$?

$\log_2$ measures the **amount of information** (in bits).

Lower probability events carry **more information**, so they contribute more to entropy.

---

## Why the Negative Sign?

Since

$$
\log_2(p)<0
\quad \text{for }0<p<1,
$$

the negative sign makes entropy **always non-negative**.

---

# Example 1: Pure Node

Dataset:

```text
[0 Yes, 3 No]
```

Probabilities:

$$
P(\text{Yes})=0
$$

$$
P(\text{No})=1
$$

Entropy:

$$
E(S)
=
-(0\log_20+1\log_21)
$$

$$
=-(0+1\times0)
$$

$$
=0
$$

**Entropy = 0 (Pure)**

---

# Example 2: Maximum Impurity

Dataset:

```text
[3 Yes, 3 No]
```

Probabilities:

$$
P(\text{Yes})=0.5
$$

$$
P(\text{No})=0.5
$$

Entropy:

$$
E(S)
=
-(0.5\log_20.5+0.5\log_20.5)
$$

Since

$$
\log_2(0.5)=-1
$$

$$
E(S)
=
-(0.5(-1)+0.5(-1))
$$

$$
=1
$$

**Entropy = 1 (Maximum impurity)**

---

# Example 3

Dataset:

```text
[3 Yes, 6 No]
```

Probabilities:

$$
P(\text{Yes})=\frac{3}{9}=0.333
$$

$$
P(\text{No})=\frac{6}{9}=0.667
$$

Entropy:

$$
E(S)
=
-\left(0.333\log_20.333
+
0.667\log_20.667\right)
$$

$$
E(S)\approx0.918
$$

---

- At $p_i=0$:

$$

$$

- At $p_i=1$:

$$

$$

$$

$$

(Maximum impurity)

---

# Entropy Range

For binary classification:

$$
0 \le E(S) \le 1
$$

| Entropy | Meaning |
|---------:|---------|
| 0 | Completely pure |
| 1 | Maximum impurity |

---

# How Decision Trees Use Entropy

For every possible feature, the tree:

1. Splits the data.
2. Computes the entropy of each child node.
3. Computes the weighted average entropy after the split.
4. Chooses the split with the **lowest entropy** (highest Information Gain).

---

# Key Takeaways

- Entropy measures **uncertainty** or **impurity**.
- Lower entropy = Better split.
- Higher entropy = More mixed classes.
- Pure node → Entropy = **0**.
- Equal class distribution → Entropy = **1**.
- Decision Trees always try to **minimize entropy**.
---
---

# Gini Impurity

Gini Impurity measures **how likely a randomly chosen sample would be classified incorrectly** if it were assigned a label according to the class distribution of a node.

Like Entropy, it measures the **impurity** of a node.

Decision Trees choose the split that **minimizes Gini Impurity**.

---

# Intuition

Think of Gini Impurity as the **probability of making a wrong prediction**.

- All samples belong to one class → No mistakes → Gini = 0
- Classes are evenly mixed → Maximum mistakes → Highest Gini

Example:

```text
[Yes, Yes, Yes, Yes]
```

Gini = **0** (Pure)

```text
[Yes, No, Yes, No]
```

Gini = **0.5** (Maximum impurity for binary classification)

---

# Gini Impurity Formula

For **c classes**:

$$
GI = 1 - \sum_{i=1}^{c} p_i^2
$$

For **binary classification**:

$$
GI = 1 - \left(P(\text{Yes})^2 + P(\text{No})^2\right)
$$

---

# Understanding the Formula

## $p_i$

Probability of class $i$.

$$
p_i=\frac{\text{Number of samples in class }i}{\text{Total number of samples}}
$$

The probabilities always satisfy:

$$
\sum p_i = 1
$$

---

# Why Are the Probabilities Squared?

Squaring gives **more importance to dominant classes**.

If one class has a very high probability, its square becomes much larger.

Example:

$$
0.9^2=0.81
$$

$$
0.1^2=0.01
$$

Their sum becomes:

$$
0.82
$$

Therefore,

$$
GI=1-0.82=0.18
$$

A dominant class produces a **low Gini Impurity**.

---

# Why Subtract from 1?

The squared probabilities measure **purity**.

Subtracting from 1 converts purity into **impurity**.

- High purity → Low Gini
- Low purity → High Gini

---

# Example 1: Pure Node

Dataset:

```text
[0 Yes, 3 No]
```

Probabilities:

$$
P(\text{Yes})=0
$$

$$
P(\text{No})=1
$$

Calculation:

$$
GI
=
1-(0^2+1^2)
$$

$$
=1-(0+1)
$$

$$
=0
$$

**Gini = 0 (Pure Node)**

---

# Example 2: Maximum Impurity

Dataset:

```text
[3 Yes, 3 No]
```

Probabilities:

$$
P(\text{Yes})=0.5
$$

$$
P(\text{No})=0.5
$$

Calculation:

$$
GI
=
1-(0.5^2+0.5^2)
$$

$$
=1-(0.25+0.25)
$$

$$
=0.5
$$

**Maximum Gini for binary classification = 0.5**

---

# Example 3

Dataset:

```text
[3 Yes, 6 No]
```

Probabilities:

$$
P(\text{Yes})=\frac{3}{9}=0.333
$$

$$
P(\text{No})=\frac{6}{9}=0.667
$$

Calculation:

$$
GI
=
1-(0.333^2+0.667^2)
$$

$$
=1-(0.111+0.444)
$$

$$
=0.445
$$

---

- **Y-axis:** Gini Impurity

### At $p=0$

Only one class exists.

$$
GI=0
$$

---

### At $p=1$

Again, only one class exists.

$$
GI=0
$$

---

### At $p=0.5$

Both classes are equally likely.

$$
GI=0.5
$$

Maximum impurity.

---

# Gini Range

For binary classification:

$$
0 \le GI \le 0.5
$$

| Gini | Meaning |
|------:|---------|
| 0 | Completely pure |
| 0.2 | Mostly one class |
| 0.4 | Highly mixed |
| 0.5 | Maximum impurity |

---

# Hidden Intuition

Gini can be interpreted as:

> **The probability of incorrectly classifying a randomly selected sample if labels are assigned according to the node's class distribution.**

Example:

```text
70 Yes
30 No
```

If you randomly assign labels based on these probabilities, there is still a chance of making a wrong prediction.

That probability is the Gini Impurity.

---



# Key Takeaways

- Gini Impurity measures **how mixed a node is**.
- Lower Gini = Better (purer) node.
- A pure node has **GI = 0**.
- A perfectly balanced binary node has **GI = 0.5**.
- Decision Trees choose the split that **minimizes the weighted Gini Impurity**.

# Entropy vs Gini Impurity

Both **Entropy** and **Gini Impurity** measure the impurity of a node in a Decision Tree.

Their objective is the same:

> **Find the split that creates the purest child nodes.**

The only difference is **how impurity is calculated**.

---

# Comparison

| Scenario | Entropy | Gini Impurity |
|----------|---------|---------------|
| Formula | $-\sum p_i\log_2(p_i)$ | $1-\sum p_i^2$ |
| Based On | Information Theory | Probability of Misclassification |
| Maximum (Binary) | 1 | 0.5 |
| Minimum | 0 | 0 |
| Computation | Uses logarithms | Uses only multiplication |
| Speed | Slightly slower | Faster |
| Split Preference | Produces slightly more balanced splits | Quickly isolates dominant classes |
| Sensitivity | More sensitive to probability differences | Less sensitive |
| Default In | ID3, C4.5 | CART, Scikit-learn |

---

# Training Speed

### Entropy

Entropy requires computing logarithms.

$$
E(S)=-\sum p_i\log_2(p_i)
$$

Logarithm calculations are computationally expensive.

Training is therefore **slightly slower**.

---

### Gini Impurity

Gini only squares probabilities.

$$
GI=1-\sum p_i^2
$$

Squaring numbers is much faster than computing logarithms.

Training is therefore **slightly faster**.

---

# Split Behaviour

Both choose the split with the lowest impurity.

However,

### Entropy

Tends to create **more balanced splits** because it reacts more strongly to small probability changes.

Example:

```text
55% Yes
45% No
```

Entropy notices this imbalance more than Gini.

---

### Gini

Gini prefers quickly separating the dominant class.

If one class already dominates,

```text
90% Yes
10% No
```

Gini becomes very small and may stop splitting earlier.

---

# Sensitivity to Class Distribution

Entropy changes more rapidly when probabilities change.

Example:

```text
50% / 50%
```

↓

```text
60% / 40%
```

Entropy decreases noticeably.

Gini also decreases, but by a smaller amount.

This makes Entropy **more sensitive**.

---

# Dataset Size

### Small datasets

Entropy is often preferred because its higher sensitivity can produce slightly better splits.

---

### Large datasets

Gini is preferred because it trains faster.

When millions of samples are involved, avoiding logarithms saves computation time.

---

# Computational Complexity

Ignoring constants,

Both have

$$
O(c)
$$

where

$c$ = number of classes.

The difference is the operation itself.

Entropy performs

- logarithms
- multiplication

Gini performs only

- multiplication
- addition

Therefore,

$$
Gini < Entropy
$$

in computation time.

---

# Graph Comparison

Both curves have the same general shape.

- Both start at **0**.
- Both end at **0**.
- Both reach their maximum when

$$
P=0.5
$$

However,

Entropy's curve is **steeper**.

Gini's curve is **flatter**.

This is why Entropy reacts more strongly to probability changes.

---

# Which One Produces Better Trees?

In practice,

**Almost no difference.**

Many studies have shown both usually produce nearly identical Decision Trees.

Prediction accuracy is often within fractions of a percent.

---

# Why Does Scikit-learn Use Gini by Default?

Scikit-learn implements the **CART (Classification and Regression Tree)** algorithm.

CART was originally designed using Gini Impurity.

Therefore,

```python
DecisionTreeClassifier()
```

internally uses

```python
criterion="gini"
```

unless changed.

---

# When Should You Use Entropy?

Choose Entropy when:

- Information Gain is required.
- You are studying Decision Trees academically.
- You want slightly more balanced splits.
- Dataset size is relatively small.

---

# When Should You Use Gini?

Choose Gini when:

- Training speed matters.
- Dataset is very large.
- Using Scikit-learn's default implementation.
- You need faster repeated training (e.g., Random Forests).

---

# Hidden Facts

### 1. Both usually choose the same split.

Although the formulas differ, the ranking of candidate splits is often identical.

---

### 2. Gini is a quadratic function.

Entropy is logarithmic.

Quadratic functions are cheaper to compute.

---

### 3. Entropy has stronger mathematical foundations.

Entropy comes from **Information Theory** and measures the expected information (in bits).

Gini does not have this information-theoretic interpretation.

---

### 4. Gini was introduced by Corrado Gini.

It originated in statistics as a measure of diversity and inequality before being adapted for Decision Trees.

---

### 5. Neither is universally better.

If one consistently outperformed the other, every ML library would use it exclusively.

The fact that both are still widely used means their practical performance is very similar.

---

# Rule of Thumb

| Situation | Recommended Metric |
|-----------|--------------------|
| Learning Decision Trees | Entropy |
| Academic explanations | Entropy |
| Small datasets | Entropy |
| Large datasets | Gini |
| Faster training | Gini |
| Scikit-learn default | Gini |
| Random Forests | Gini (commonly used) |

---

# Key Takeaways

- Both metrics measure **node impurity**.
- Both aim to create **pure child nodes**.
- Entropy uses logarithms and is more sensitive.
- Gini uses squared probabilities and is faster.
- Both usually produce very similar Decision Trees.
- In practice, **Gini is preferred for speed**, while **Entropy is preferred for interpretability and Information Gain**.

# Information Gain (IG)

Information Gain measures **how much uncertainty (impurity) decreases after splitting a node**.

Its purpose is to decide:

> **Which feature should the Decision Tree split on?**

The feature with the **highest Information Gain** is selected.

---

# Intuition

Before splitting, the parent node contains mixed classes.

After splitting, the child nodes should become purer.

Information Gain measures **how much impurity was removed**.

More impurity removed = Better split.

---

# Formula (Using Entropy)

$$
IG(S,A)
=
H(S)
-
\sum_{v\in A}
\frac{|S_v|}{|S|}
H(S_v)
$$

where

- $H(S)$ = Entropy of the parent node
- $H(S_v)$ = Entropy of each child node
- $\frac{|S_v|}{|S|}$ = Weight (fraction of samples in that child)
- $A$ = Feature being evaluated

---

# Formula (Using Gini)

The same idea applies using Gini Impurity.

$$
Gini\ Gain
=
GI(S)
-
\sum_{v\in A}
\frac{|S_v|}{|S|}
GI(S_v)
$$

Only the impurity measure changes.

---

# Why Do We Need Weights?

Not every child node contains the same number of samples.

Example:

Parent:

```text
100 samples
```

Split into:

```text
90 samples

10 samples
```

The larger child should influence the decision more.

Therefore,

$$
Weight
=
\frac{\text{Samples in child}}
{\text{Samples in parent}}
$$

---

# What Is Actually Happening?

For every feature,

the Decision Tree performs these steps:

1. Calculate impurity of the parent node.
2. Split the data using that feature.
3. Calculate impurity of every child.
4. Compute the **weighted average impurity**.
5. Calculate Information Gain.

Repeat this for **every feature**.

Finally,

choose the feature with the **highest Information Gain**.

---

# Why Subtract?

Before splitting,

the node has some uncertainty.

After splitting,

the uncertainty should decrease.

So,

$$
Information\ Gain
=
Before
-
After
$$

The larger the reduction,

the better the feature.

---

# Example

Parent Entropy:

$$
H(S)=0.918
$$

After splitting:

| Child | Weight | Entropy |
|-------|--------|---------|
| Left | 3/6 | 0.918 |
| Right | 3/6 | 0 |

Weighted Entropy:

$$
\frac{3}{6}(0.918)
+
\frac{3}{6}(0)
=
0.459
$$

Information Gain:

$$
IG
=
0.918
-
0.459
=
0.459
$$

---

# Which Information Gain Is Better?

Always choose the **highest Information Gain**.

Example:

| Feature | Information Gain |
|---------|------------------|
| Age | 0.18 |
| Salary | 0.52 |
| Credit Score | 0.31 |

The tree chooses

```text
Salary
```

because

```text
0.52
```

is the largest reduction in impurity.

---

# Range of Information Gain

Minimum:

$$
IG=0
$$

Meaning:

The split did not improve purity.

---

Maximum:

Occurs when the split makes every child node perfectly pure.

Higher Information Gain is always better.

---

# Greedy Nature of Decision Trees

Decision Trees use a **Greedy Algorithm**.

A greedy algorithm makes the **best decision at the current step** without considering future splits.

Example:

At the root node,

the tree computes Information Gain for every feature.

```text
Age           → 0.25

Salary        → 0.41

Credit Score  → 0.63
```

The tree immediately selects

```text
Credit Score
```

because it has the highest Information Gain.

It **does not check** whether choosing another feature now could produce a better overall tree later.

---

# Hidden Insight

Information Gain is **not** the impurity itself.

It is the **reduction in impurity**.

Think of it as:

```text
Before Split
      ↓
Remove uncertainty
      ↓
After Split

Information Gain = Uncertainty Removed
```

---

# Entropy vs Gini in Information Gain

Both follow the same process.

The only difference is the impurity measure used.

Using Entropy:

$$
IG
=
Parent\ Entropy
-
Weighted\ Child\ Entropy
$$

Using Gini:

$$
Gini\ Gain
=
Parent\ Gini
-
Weighted\ Child\ Gini
$$

Both choose the split with the **largest impurity reduction**.

---

# Key Takeaways

- Information Gain measures **how much impurity decreases after a split**.
- It is computed for **every feature**.
- Child impurities are **weighted** by the number of samples.
- The feature with the **highest Information Gain** is selected.
- Decision Trees use a **greedy approach**, making the best local split at each node.
- Information Gain can be computed using either **Entropy** or **Gini Impurity**.

# How Are Child Nodes Created Before Selecting the Best Feature?

The `child nodes` inside the Information Gain formula are `temporary candidate child nodes`, not real tree nodes yet.  
The tree **does not permanently create child nodes** for every feature.

Instead, it **temporarily simulates** a split using each feature, calculates the impurity reduction (Information Gain), and then discards it.

Example:

```text
Try Age
→ Calculate IG
→ Discard

Try Salary
→ Calculate IG
→ Discard

Try Credit Score
→ Calculate IG
→ Discard
```

The feature with the **highest Information Gain** is finally selected, and **only then** are the child nodes actually created.

> The child nodes used in the Information Gain formula are **temporary candidate nodes**, not permanent tree nodes.

---

# Is Information Gain Calculated at Every Split?

**Yes.**

At **every node**, the Decision Tree repeats the same process:

1. Try every feature.
2. Simulate the split.
3. Calculate Information Gain.
4. Choose the feature with the highest Information Gain.
5. Create the child nodes.
6. Repeat for each child node.

This continues until a stopping condition is reached.

---

# Doesn't This Increase Time Complexity?

Yes.

Training a Decision Tree is computationally expensive because Information Gain is calculated repeatedly at every node.

However:

- **Training happens only once.**
- **Prediction is very fast**, since it only follows one path from the root to a leaf without recalculating Information Gain.

---

# Why Is It Called a Greedy Algorithm?

A Decision Tree is **greedy** because it always chooses the **best split at the current node** (highest Information Gain).

It **does not** look ahead to see whether another choice could produce a better overall tree.

It makes the best **local decision**, then moves to the next node.

# Pruning

Pruning is the process of **removing unnecessary branches** from a Decision Tree to reduce **overfitting** and improve performance on unseen data.

The goal is to build a **simpler and more generalized** tree.

---

# Why Do We Need Pruning?

A Decision Tree can keep splitting until every leaf becomes pure.

This creates a very complex tree that **memorizes the training data**.

This is called **overfitting**.

```text
Training Accuracy  → High

Testing Accuracy   → Low
```

Pruning removes branches that contribute little to prediction accuracy.

---

# Intuition

Imagine memorizing every question from a previous exam.

You'll score well on that exam but struggle on a new one.

Instead, learning the concepts helps you solve unseen questions.

Pruning does the same:

- Removes memorization.
- Improves generalization.

---

# Analogy

Think of pruning a real tree.

A gardener removes weak or unnecessary branches so the tree grows healthier.

Similarly, Decision Tree pruning removes unnecessary branches so the model predicts better.

---

# When Is Pruning Used?

Pruning is used when the tree:

- Becomes very deep.
- Starts memorizing noise.
- Shows signs of overfitting.

---

# Types of Pruning

## 1. Pre-Pruning (Early Stopping)

The tree is **stopped before it becomes too large**.

Common stopping rules:

- `max_depth`
- `min_samples_split`
- `min_samples_leaf`
- `max_leaf_nodes`
- `min_impurity_decrease`

---

## 2. Post-Pruning

The tree is first grown completely.

Then unnecessary branches are removed.

A common method is:

- **Cost Complexity Pruning (ccp_alpha)**

---

# Pre-Pruning vs Post-Pruning

| Pre-Pruning | Post-Pruning |
|-------------|--------------|
| Stops tree growth early | Grows full tree first |
| Faster training | Slightly slower |
| May underfit if stopped too early | Usually produces better generalization |

---

# Common Pruning Rules

## `max_depth`

Maximum depth allowed.

```python
max_depth = 3
```

The tree cannot grow beyond 3 levels.

---

## `min_samples_split`

Minimum number of samples required to split a node.

```python
min_samples_split = 20
```

Nodes with fewer than 20 samples are not split.

---

## `min_samples_leaf`

Minimum number of samples required in each leaf.

```python
min_samples_leaf = 5
```

Every leaf must contain at least 5 samples.

---

## `max_leaf_nodes`

Limits the total number of leaf nodes.

Prevents the tree from becoming too large.

---

## `min_impurity_decrease`

A split is performed only if it reduces impurity by at least a threshold.

Example:

```python
min_impurity_decrease = 0.01
```

Tiny improvements are ignored.

---

## `ccp_alpha` (Cost Complexity Pruning)

Used for **post-pruning**.

Higher `ccp_alpha` removes more branches.

- Small value → Larger tree
- Large value → Smaller tree

---

# Hidden Intuition

Not every split is useful.

Some branches reduce impurity only slightly but greatly increase tree complexity.

Pruning removes these branches because the small improvement is not worth the added complexity.

---

# Key Takeaways

- Pruning reduces **overfitting**.
- It removes unnecessary branches.
- The goal is better **generalization** on unseen data.
- **Pre-pruning** stops growth early.
- **Post-pruning** removes branches after the full tree is built.
- Common pruning rules include:
  - `max_depth`
  - `min_samples_split`
  - `min_samples_leaf`
  - `max_leaf_nodes`
  - `min_impurity_decrease`
  - `ccp_alpha`

# Decision Tree Classification: Step-by-Step

1. **Start with the entire dataset.**

2. **Calculate the impurity** (Entropy or Gini) of the current node.

3. **Try every feature and every possible split.**

4. **For each split:**
   - Split the data into child nodes.
   - Calculate the impurity of each child.
   - Compute the **weighted child impurity**.
   - Calculate the **Information Gain (or Gini Gain)**.

5. **Choose the split with the highest Information Gain** (or highest Gini Gain).

6. **Create the child nodes** using that best split.

7. **Repeat recursively** for each child node until a stopping condition is met (e.g., `max_depth`, `min_samples_split`, etc.).

8. **At each leaf node, store the majority class** (the class with the highest number of samples).

9. **Prediction:**  
   For a new sample, follow the decision rules from the root to a leaf and **return the majority class** stored in that leaf.

# Decision Tree for Regression

Unlike **Decision Tree Classification**, which predicts a class, a **Decision Tree Regressor** predicts a **continuous numerical value**.

Instead of using **Entropy** or **Gini Impurity**, it uses **Variance Reduction (VR)** to find the best split.

---

# Goal

The objective is to split the data so that the target values within each child node become as similar as possible.

Lower variance = Better split.

---

# Overall Process

1. Consider every feature.
2. Try every possible split.
3. Compute the **Variance Reduction** for each split.
4. Choose the split with the **highest Variance Reduction**.
5. Split the data.
6. Repeat recursively for each child node.
7. Stop when a stopping criterion is met.
8. Store the **average target value** in each leaf.
9. During prediction, return that average.

---

# Why Variance?

In regression, outputs are numbers.

Instead of measuring class impurity, we measure how spread out the numerical values are.

Example:

```text
[5, 5, 5, 5]
```

Variance = 0 (perfect)

---

```text
[2, 4, 6, 10]
```

Variance is high.

The tree tries to reduce this variance after every split.

---

# Variance Formula

$$
Var(S)
=
\frac1n
\sum_{i=1}^{n}
(y_i-\bar y)^2
$$

where

- $y_i$ = target value
- $\bar y$ = mean of target values
- $n$ = number of samples

---
# Variance Reduction Formula

$$
VR
=
V(S)
-
\left(
\frac{|S_L|}{|S|}V(S_L)
+
\frac{|S_R|}{|S|}V(S_R)
\right)
$$

where

- $V(S)$ = Variance of the parent node
- $V(S_L)$ = Variance of the left child
- $V(S_R)$ = Variance of the right child
- $|S|$ = Number of samples in the parent node
- $|S_L|$ = Number of samples in the left child
- $|S_R|$ = Number of samples in the right child

---

# What Does the Formula Mean?

The formula calculates:

> **Parent Variance − Weighted Average Variance of the Child Nodes**

The larger the reduction, the better the split.

---

# Steps

1. Calculate the variance of the parent node.
2. Split the data into left and right child nodes.
3. Calculate the variance of each child.
4. Compute the weighted average variance of the children.
5. Subtract it from the parent variance.

The result is the **Variance Reduction (VR)**.

---

Compare both Variance Reductions.

| Split | Variance Reduction |
|--------|-------------------:|

---

# Why Highest Variance Reduction?

Higher Variance Reduction means

- child nodes are more similar
- predictions become more accurate
- error decreases

Exactly like Information Gain in classification,

Regression chooses the split that removes the **most variance**.

---

# Recursive Process

After selecting the best split,

the algorithm repeats the same steps on each child node:

1. Calculate variance.
2. Try every possible split.
3. Compute Variance Reduction.
4. Select the best split.
5. Repeat until stopping criteria are met.

---

# When Does It Stop?

The tree stops growing when:

- Maximum depth is reached.
- Minimum samples required for a split are not available.
- Variance Reduction is too small.
- Node becomes sufficiently pure.

---

# Prediction

Once training is complete,

each **leaf node stores the average (mean)** of all target values inside it.

Example:

Leaf contains

```text
[2,4]
```

Prediction

$$
\frac{2+4}{2}
=
3
$$

---

Another leaf

```text
[8,10,12]
```

Prediction

$$
\frac{8+10+12}{3}
=
10
$$

---

For a new sample,

the tree simply follows the decision rules until it reaches a leaf and returns that stored average.

---

# Why Do We Predict the Mean?

The mean minimizes the **Mean Squared Error (MSE)** within a node.

Among all possible prediction values,

the average gives the smallest squared error.

This is why Decision Tree Regressors store the mean at every leaf.

---

# Classification vs Regression

| Classification | Regression |
|---------------|------------|
| Predicts a class | Predicts a number |
| Entropy / Gini | Variance |
| Information Gain | Variance Reduction |
| Leaf stores majority class | Leaf stores mean target value |

---

# Do Decision Tree Regressors Need Feature Scaling?

**No.**

Decision Trees split using simple comparisons such as

```text
Age ≤ 30

Fare > 50

Salary ≤ 70000
```

The exact numerical scale does not affect these comparisons.

For example,

```text
Age = 25
```

or

```text
Age = 0.25
```

(after scaling)

will produce equivalent split decisions because the threshold changes accordingly.

Since the algorithm never computes distances or gradients, scaling has no impact on how splits are chosen.

---

# Key Takeaways

- Regression trees predict **continuous values**.
- They use **Variance Reduction**, not Entropy or Gini.
- Every possible split is evaluated.
- The split with the **highest Variance Reduction** is selected.
- The process repeats recursively until stopping conditions are met.
- Each leaf stores the **mean target value**.
- Predictions are the **average of the values in the reached leaf**.
- Feature scaling is **not required** because decisions are based only on threshold comparisons.
