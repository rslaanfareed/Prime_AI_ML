```python
import json
```

```python
# load data
def load_data(filename):
    with open(filename,"r") as f:
        data=json.load(f)
    return data
```

```python
data=load_data("customer_data.json")
```

```python
# clean data
def clean_data(data):
    cleaned_data=[]
    unique_customers=set()
    for customer in data:

        #city to lowercase and add missing values
        raw_city=customer.get("city")
        if (raw_city is None) or (raw_city.strip()==""):
            customer["city"]="unknown"
        
        else:
            customer["city"]=raw_city.strip().lower()
    
        #payment method to lowercase and strip the spaces
        raw_payment=str(customer.get("payment"))
        customer["payment"]=raw_payment.strip().lower()
    
        # amount N/A to None and convert to integer whenever possible
        raw_amount=customer.get("amount")
        if raw_amount== "N/A":
            customer["amount"]=None
        elif raw_amount is not None:
            try:
                customer["amount"]=int(raw_amount)
            except ValueError:
                pass

        customer_name=customer["customer"].strip()
        if customer_name in unique_customers:
            continue
        
        unique_customers.add(customer_name)
        customer["customer"]=customer_name
        cleaned_data.append(customer)
    
    return cleaned_data
    
    
```

```python
data=clean_data(data)
```

```python
# get meaningful insights

def get_insights(data):

    #Average order amount
    amount_sum=0
    for customer in data:
        if customer.get("amount") is not None:
            amount_sum+=int(customer.get("amount"))
    print(f"Average order amount: {amount_sum/len(data):.2f}")

    #highest order amount
    highest=0
    for customer in data:
        if customer.get("amount") is not None:
            if customer.get("amount")>highest:
                highest=customer.get("amount")
    print(f"Highest order amount: {highest}")

    #lowest order amount
    lowest=10000000
    for customer in data:
        if customer.get("amount") is not None:
            if customer.get("amount")<lowest:
                lowest=customer.get("amount")
    print(f"lowest order amount: {lowest}")

    #Number of delivered orders, cancelled orders and total sales
    delivered_count=0
    cancelled_count=0
    total_sales=0
    for customer in data:
        if customer.get("amount") is not None:
            total_sales+=customer.get("amount")
        if customer.get("status").strip().lower()=="delivered":
            delivered_count+=1
        elif customer.get("status").strip().lower()=="cancelled":
            cancelled_count+=1
    print(f"Delivered orders: {delivered_count}")
    print(f"Cancelled orders: {cancelled_count}")
    print(f"Total sales: {total_sales}")

    #which city has how many customers
    lhr_count=0
    isl_count=0
    khi_count=0
    pesh_count=0
    for customer in data:
        if customer.get("city")=="islamabad":
            isl_count+=1
        elif customer.get("city")=="lahore":
            lhr_count+=1
        elif customer.get("city")=="karachi":
            khi_count+=1
        elif customer.get("city")=="peshawar":
            pesh_count+=1    
    
    print(f"Customers from Islamabad: {isl_count}")
    print(f"Customers from Lahore: {lhr_count}")
    print(f"Customers from Karachi: {khi_count}")
    print(f"Customers from Peshawar: {pesh_count}")

    #cash vs card percentage
    cash_cust=0
    card_cust=0
    for cust in data:
        if cust.get("payment").strip().lower()=="cash":
            cash_cust+=1
        elif cust.get("payment").strip().lower()=="card":
            card_cust+=1
    print(f"% of cash customers: {cash_cust/len(data)*100:.2f}")
    print(f"% of card customers: {card_cust/len(data)*100:.2f}")
    
            

    

    
    
```

```python
get_insights(data)
```
