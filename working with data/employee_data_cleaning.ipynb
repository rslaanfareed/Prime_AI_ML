```python
import json
```

```python
#load data
def load_data(filename):
    with open(filename,"r") as f:
        data=json.load(f)
    return data
```

```python
data=load_data("emp_data.json")
```

```python
#data cleaning
def data_cleaning(data):

    unique_emps=set()
    cleaned_data=[]
    # text to number for experience values
    text_to_num={"one":1,"two":2,"three":3,"four":4,"five":5,"six":6,"seven":7,"eight":8,"nine":9}

    for emp in data:
        
        #name values strip and lower
        raw_name=emp.get("name")
        emp["name"]=raw_name.strip().lower()

        #standardize department values
        raw_dep=emp.get("department")
        emp["department"]=raw_dep.strip().lower()

        #salary to integers
        raw_sal=str(emp.get("salary"))
        if (raw_sal is None or raw_sal.strip()==""):
            emp["salary"]=45000
        else:
            emp["salary"]=int(raw_sal)

        #experience to integer
        if emp.get("experience") is not None:
            raw_exp=str(emp.get("experience"))
            if raw_exp in text_to_num:
                emp["experience"]=int(text_to_num[raw_exp])
        else:
            emp["experience"]=0

        if emp["name"].strip() in unique_emps:
            continue
        unique_emps.add(emp["name"])
        cleaned_data.append(emp)
    return cleaned_data
        
```

```python
data=data_cleaning(data)
```

```python
#useful insights

def get_insights(data):
    
    sal_sum=0
    
    for emp in data:
        #salary sum to calculate average
        sal_sum+=emp.get("salary")
    #print average salary
    print(f"Average salary: {sal_sum/len(data)}\n")

    #highest paid employee
    highest=0
    for emp in data:
        if emp.get("salary")>highest:
            highest=emp.get("salary")
            highest_paid_emp=emp["name"]
    #print highest employee
    print(f"Highest paid employee: {highest_paid_emp}")
    print(f"salary: {highest}\n")
        
    #lowest paid employee
    lowest=999999999
    for emp in data:
        if emp.get("salary")<lowest:
            lowest=emp.get("salary")
            lowest_paid_emp=emp["name"]
    #print lowest employee
    print(f"lowest paid employee: {lowest_paid_emp}")
    print(f"salary: {lowest}\n")

    #department highest average salary
    
    #finance avg
    hr_sum=0
    hr_emp_count=0
    fin_sum=0
    fin_emp_count=0
    it_sum=0
    it_emp_count=0

    for emp in data:
        if emp.get("department").strip().lower()=="finance":
            fin_sum+=emp.get("salary")
            fin_emp_count+=1

        elif emp.get("department").strip().lower()=="it":
            it_sum+=emp.get("salary")
            it_emp_count+=1

        elif emp.get("department").strip().lower()=="hr":
            hr_sum+=emp.get("salary")
            hr_emp_count+=1

    # Store averages in a dictionary
    averages = {
        "HR": hr_sum/hr_emp_count,
        "IT": it_sum/it_emp_count,
        "Finance": fin_sum/fin_emp_count
    }

    # Find the key with the maximum value
    highest_dept = max(averages, key=averages.get)
    highest_salary = averages[highest_dept]
    
    print(f"Highest average salary department: {highest_dept} , Average salary: {highest_salary}\n")

    #Number of emp in each department
    print(f"HR department: {hr_emp_count} employees")
    print(f"IT department: {it_emp_count} employees")
    print(f"Finance department: {fin_emp_count} employees\n")
    
    #average experience
    exp_sum=0
    for emp in data:
        exp_sum+=int(emp.get("experience"))
    print(f"Average Experience: {exp_sum/len(data)}\n")

    #excellent performance percentage
    exc_count=0
    for emp in data:
        if emp.get("performance").strip().lower()=="excellent":
            exc_count+=1
    print(f"Excellent performers percentage: {exc_count/len(data)*100:.2f}")
            
            
        
            
            
    
```

```python
get_insights(data)
```

```python
#recommendations
def get_recommendations(data):
    recommendations=[]
    for emp in data:
        curr_rec={"name":emp.get("name")}
        
        if emp.get("performance").strip().lower()=="excellent":
            curr_rec["recommendation"]="promotion"
        elif int(emp.get("experience"))<=2:
            curr_rec["recommendation"]="training"

        elif emp.get("performance").strip().lower()=="poor":
            curr_rec["recommendation"]="performance review"
        else:
            curr_rec["recommendation"]="No recommendation"
        recommendations.append(curr_rec)
    print(recommendations)
            
```

```python
get_recommendations(data)
```
